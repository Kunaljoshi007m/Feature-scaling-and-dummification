*  Perform feature dummification to convert categorical variables into numerical representations.







#Part 1: Handling Numerical Data (Setup)

```python
import numpy as np
import pandas as pd
from sklearn import preprocessing
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans

# Sample Data from Heart Disease dataset
data = pd.read_csv("heart.csv")

df = pd.DataFrame(data)
print("Original Data:\n", df.head())
```

#2: Rescaling a Feature (MinMax Scaling)

```python
feature = df[['age']].values

minmax_scaler = preprocessing.MinMaxScaler(feature_range=(0, 1))
scaled_feature = minmax_scaler.fit_transform(feature)

print("\nScaled Age (0-1):\n", scaled_feature)
```

#3: Standardizing a Feature (Z-Score & Robust)

```python
# A. Standard Scaler (Z-Score) on 'trestbps'
feature_bps = df[['trestbps']].values
scaler = preprocessing.StandardScaler()
standardized = scaler.fit_transform(feature_bps)

print("\nStandardized trestbps:\n", standardized)
print("Mean:", round(standardized.mean()))
print("Std Dev:", standardized.std())

# B. Robust Scaler on 'chol' (Good for outliers)
feature_chol = df[['chol']].values
robust_scaler = preprocessing.RobustScaler()
robust_chol = robust_scaler.fit_transform(feature_chol)

print("\nRobust Scaled Chol:\n", robust_chol)
```

#4: Normalizing Observations

```python
# Target: Combining 'age' and 'thalach' into a vector
features_vector = df[['age', 'thalach']].values.astype(float)

# L2 Normalization (Euclidean norm)
normalizer_l2 = preprocessing.Normalizer(norm="l2")
print("\nL2 Normalization:\n", normalizer_l2.transform(features_vector))

# Max Normalization
normalizer_max = preprocessing.Normalizer(norm="max")
print("\nMax Normalization:\n", normalizer_max.transform(features_vector))
```

#5: Grouping Observations Using Clustering

```python
# Grouping patients based on 'age' and 'chol'
features_cluster = df[['age', 'chol']]

# Create 2 clusters
clusterer = KMeans(n_clusters=2, random_state=0)
clusterer.fit(features_cluster)

# Assign the group to the dataframe
df['group'] = clusterer.predict(features_cluster)

print("\nData with Cluster Groups:\n", df[['age', 'chol', 'group']])
```

#6: Handling Missing Numerical Values

```python
# Setup: Creating a numpy array with NaNs based on 'oldpeak' and 'chol'
features_missing = np.array([
    [1.0, 212],
    [3.1, 203],
    [np.nan, 174], # Missing oldpeak
    [0.0, np.nan], # Missing chol
    [1.9, 294]
])

print("\nOriginal Features with NaN:\n", features_missing)

# A. Drop Observations with NaN
clean_features = features_missing[~np.isnan(features_missing).any(axis=1)]
print("\nDropped Rows with NaN:\n", clean_features)

# B. Impute with Mean
mean_imputer = SimpleImputer(strategy="mean")
features_mean = mean_imputer.fit_transform(features_missing)
print("\nImputed with Mean:\n", features_mean)

# C. Impute with Median
median_imputer = SimpleImputer(strategy="median")
features_median = median_imputer.fit_transform(features_missing)
print("\nImputed with Median:\n", features_median)
```
